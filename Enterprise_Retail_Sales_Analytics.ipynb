```python
query = """
SELECT
    Category,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM retail_sales
GROUP BY Category
ORDER BY Total_Sales DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Region,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM retail_sales
GROUP BY Region
ORDER BY Total_Sales DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Category,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM retail_sales
GROUP BY Category
HAVING SUM(Profit) > 50000
ORDER BY Total_Profit DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Discount,
    COUNT(*) AS Total_Orders,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit,
    ROUND(AVG(Profit), 2) AS Average_Profit
FROM retail_sales
GROUP BY Discount
ORDER BY Discount;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Category,
    COUNT(*) AS Loss_Records,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Loss
FROM retail_sales
WHERE Profit < 0
GROUP BY Category
ORDER BY Total_Loss ASC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    CASE
        WHEN Discount <= 0.20 THEN 'Low Discount'
        WHEN Discount <= 0.40 THEN 'Medium Discount'
        ELSE 'High Discount'
    END AS Discount_Category,

    COUNT(*) AS Total_Records,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit

FROM retail_sales
GROUP BY Discount_Category
ORDER BY Total_Profit DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Sub_Category,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM retail_sales
GROUP BY Sub_Category
ORDER BY Total_Profit DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Sub_Category,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM retail_sales
GROUP BY Sub_Category
HAVING SUM(Profit) < 0
ORDER BY Total_Profit ASC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Ship_Mode,
    COUNT(*) AS Total_Records,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit,
    ROUND(AVG(Profit), 2) AS Average_Profit
FROM retail_sales
GROUP BY Ship_Mode
ORDER BY Total_Profit DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Region,
    Category,
    SUM(Sales) AS Total_Sales,

    RANK() OVER (
        PARTITION BY Region
        ORDER BY SUM(Sales) DESC
    ) AS Sales_Rank

FROM retail_sales
GROUP BY Region, Category
ORDER BY Region, Sales_Rank;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Region,
    Category,
    SUM(Profit) AS Total_Profit,

    DENSE_RANK() OVER (
        PARTITION BY Region
        ORDER BY SUM(Profit) DESC
    ) AS Profit_Rank

FROM retail_sales
GROUP BY Region, Category
ORDER BY Region, Profit_Rank;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Category,
    Sales,
    Profit
FROM retail_sales
WHERE Sales > (
    SELECT AVG(Sales)
    FROM retail_sales
)
ORDER BY Sales DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
CREATE TABLE region_summary AS
SELECT
    Region,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM retail_sales
GROUP BY Region;
"""

conn.execute("DROP TABLE IF EXISTS region_summary")
conn.execute(query)
conn.commit()

print("Region summary table created successfully!")
```

```python
pd.read_sql_query(
    "SELECT * FROM region_summary;",
    conn
)
```

```python
query = """
SELECT
    r.Region,
    r.Category,
    r.Sales,
    r.Profit,
    s.Total_Sales AS Region_Total_Sales,
    s.Total_Profit AS Region_Total_Profit
FROM retail_sales r
INNER JOIN region_summary s
ON r.Region = s.Region
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    r.Region,
    r.Category,
    r.Sales,
    s.Total_Sales AS Region_Total_Sales
FROM retail_sales r
LEFT JOIN region_summary s
ON r.Region = s.Region
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Category,
    COUNT(*) AS Total_Records,

    SUM(Sales) AS Total_Sales,

    SUM(Profit) AS Total_Profit,

    SUM(CASE
        WHEN Profit < 0 THEN 1
        ELSE 0
    END) AS Loss_Records,

    ROUND(
        (SUM(Profit) / SUM(Sales)) * 100,
        2
    ) AS Profit_Margin_Percentage

FROM retail_sales
GROUP BY Category
ORDER BY Total_Profit DESC;
"""

pd.read_sql_query(query, conn)
```
